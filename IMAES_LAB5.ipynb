```python
# -----------------------------
# Facts already known to system
# -----------------------------
facts = set()

# -----------------------------
# Rules
# Format:
# conclusion -> list of conditions
# -----------------------------
rules = {
    "battery_fault": [
        "dim_headlights",
        "engine_not_cranking"
    ],

    "fuel_system_fault": [
        "engine_cranks",
        "engine_does_not_start"
    ],

    "starter_motor_fault": [
        "clicking_sound",
        "engine_not_cranking"
    ],

    "alternator_fault": [
        "battery_warning_light",
        "battery_discharges"
    ]
}


# -----------------------------
# Askable facts
# -----------------------------
askable_facts = {
    "dim_headlights": "Are the headlights dim? (yes/no)",
    "engine_not_cranking": "Is the engine not cranking? (yes/no)",
    "engine_cranks": "Does the engine crank? (yes/no)",
    "engine_does_not_start": "Does the engine fail to start? (yes/no)",
    "clicking_sound": "Do you hear a clicking sound when starting? (yes/no)",
    "battery_warning_light": "Is the battery warning light ON? (yes/no)",
    "battery_discharges": "Does the battery discharge quickly? (yes/no)"
}


# -----------------------------
# Find rule responsible for
# asking about a fact
# -----------------------------
def find_rule_for_fact(fact, goal):
    """
    Find the rule currently being used to prove the goal
    that requires this fact.
    """

    if goal in rules and fact in rules[goal]:
        return goal

    return None


# -----------------------------
# Ask user about a fact
# -----------------------------
def ask_user(fact, goal):
    while True:
        answer = input(
            f"\n{askable_facts[fact]}\n"
            "Enter yes/no/WHY: "
        ).strip().lower()

        if answer == "why":
            rule = find_rule_for_fact(fact, goal)

            if rule:
                conditions = " AND ".join(rules[rule])

                print(
                    f"\nWHY is this question being asked?"
                )
                print(
                    f"Because the system is checking whether "
                    f"'{rule}' is true."
                )
                print(
                    f"Rule: IF {conditions} "
                    f"THEN {rule}"
                )
            else:
                print("\nNo rule found for this question.")

        elif answer == "yes":
            facts.add(fact)
            return True

        elif answer == "no":
            return False

        else:
            print("Please enter yes, no, or WHY.")


# -----------------------------
# Backward Chaining Algorithm
# -----------------------------
def backward_chain(goal):

    # If fact is already known
    if goal in facts:
        return True

    # If goal cannot be derived using a rule
    if goal not in rules:
        return False

    # Check every condition required by the rule
    for condition in rules[goal]:

        # If condition already known, continue
        if condition in facts:
            continue

        # If condition is askable, ask user
        if condition in askable_facts:

            if ask_user(condition, goal):
                continue
            else:
                return False

        # Otherwise try to prove condition recursively
        if not backward_chain(condition):
            return False

    # All conditions are satisfied
    facts.add(goal)

    return True


# -----------------------------
# Main diagnosis function
# -----------------------------
def diagnose():

    print("===================================")
    print("   CAR FAULT DIAGNOSIS SYSTEM")
    print("   Backward Chaining")
    print("===================================")

    # Hypothesis order
    hypotheses = [
        "battery_fault",
        "starter_motor_fault",
        "fuel_system_fault",
        "alternator_fault"
    ]

    for hypothesis in hypotheses:

        print(f"\nChecking hypothesis: {hypothesis}")

        if backward_chain(hypothesis):

            print(
                f"\nDiagnosis: {hypothesis}"
            )

            return

    print("\nNo fault could be diagnosed.")


# -----------------------------
# Run program
# -----------------------------
if __name__ == "__main__":
    diagnose()
```
