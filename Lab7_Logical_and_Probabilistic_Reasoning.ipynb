### Exercise 1: Propositional Logic Truth Table
**Statement:** *"If it rains ($R$), then the road is wet ($W$)"*

**Formula:** $R \rightarrow W \equiv \neg R \lor W$

```python
# Exercise 1: Propositional Logic
print("=" * 65)
print("EXERCISE 1: PROPOSITIONAL LOGIC (Truth Table Generation)")
print("=" * 65)
print(f"{'Rain (R)':<12} | {'Wet Road (W)':<15} | {'NOT R':<10} | {'Implication (R -> W)':<20}")
print("-" * 65)

truth_values = [True, False]
for R in truth_values:
    for W in truth_values:
        not_R = not R
        implication = (not R) or W
        print(f"{str(R):<12} | {str(W):<15} | {str(not_R):<10} | {str(implication):<20}")

print("-" * 65)
print("Verification: Implication is FALSE ONLY when R is True and W is False.")
```

### Exercise 2: First-Order Logic (FOL) Inference
**Premises:**
1. Universal Rule: $\forall x \, (\text{Human}(x) \rightarrow \text{Mortal}(x))$ (*"All humans are mortal"*)
2. Fact: $\text{Human}(\text{Socrates})$ (*"Socrates is a human"*)
3. Fact: $\text{Human}(\text{Plato})$
4. Fact: $\text{God}(\text{Zeus})$

```python
# Exercise 2: First-Order Logic Inference
class PredicateKB:
    def __init__(self):
        self.facts = {}
        self.rules = []

    def add_fact(self, predicate: str, entity: str):
        if predicate not in self.facts:
            self.facts[predicate] = set()
        self.facts[predicate].add(entity)

    def add_universal_rule(self, antecedent_pred: str, consequent_pred: str):
        self.rules.append((antecedent_pred, consequent_pred))

    def query_mortal(self, entity: str) -> bool:
        if "Mortal" in self.facts and entity in self.facts["Mortal"]:
            return True
        for ant, cons in self.rules:
            if cons == "Mortal":
                if ant in self.facts and entity in self.facts[ant]:
                    self.add_fact("Mortal", entity)
                    return True
        return False

print("=" * 65)
print("EXERCISE 2: FIRST-ORDER LOGIC INFERENCE")
print("=" * 65)

kb = PredicateKB()
kb.add_universal_rule("Human", "Mortal")
kb.add_fact("Human", "Socrates")
kb.add_fact("Human", "Plato")
kb.add_fact("God", "Zeus")

for entity in ["Socrates", "Plato", "Zeus"]:
    print(f"Query: Is {entity} mortal? -> {kb.query_mortal(entity)}")
```

### Exercises 3 & 4: Probability of Events & Conditional Probability
**Dataset:** Contingency Table of 100 Students (Attendance vs Performance)

```python
# Exercises 3 & 4: Probabilities
N = 100
n_A_and_P = 60
n_A_and_notP = 10
n_notA_and_P = 10
n_notA_and_notP = 20

n_A = n_A_and_P + n_A_and_notP  # 70
n_P = n_A_and_P + n_notA_and_P  # 70
n_notA = N - n_A                 # 30

# Exercise 3
P_A = n_A / N
P_P = n_P / N
P_A_cap_P = n_A_and_P / N
P_A_cup_P = P_A + P_P - P_A_cap_P

print("=" * 65)
print("EXERCISE 3: INDIVIDUAL, JOINT & UNION PROBABILITIES")
print("=" * 65)
print(f"P(A) [Attending]             = {P_A:.4f}")
print(f"P(P) [Passing]               = {P_P:.4f}")
print(f"P(A ∩ P) [Attended & Passed] = {P_A_cap_P:.4f}")
print(f"P(A ∪ P) [Attended OR Passed]= {P_A_cup_P:.4f}")

# Exercise 4
P_P_given_A = P_A_cap_P / P_A
P_A_given_P = P_A_cap_P / P_P
P_P_given_notA = (n_notA_and_P / N) / (n_notA / N)

print("\n" + "=" * 65)
print("EXERCISE 4: CONDITIONAL PROBABILITIES")
print("=" * 65)
print(f"P(Pass | Attended)      = {P_P_given_A:.4f} ({P_P_given_A*100:.2f}%)")
print(f"P(Attended | Pass)      = {P_A_given_P:.4f} ({P_A_given_P*100:.2f}%)")
print(f"P(Pass | Did Not Attend) = {P_P_given_notA:.4f} ({P_P_given_notA*100:.2f}%)")
```

### Exercises 5 & 6: Bayes' Theorem & Visualization
**Bayes' Formula:**
$$P(D \mid +) = \frac{P(+ \mid D) \cdot P(D)}{P(+ \mid D) \cdot P(D) + P(+ \mid \neg D) \cdot P(\neg D)}$$

```python
# Exercises 5 & 6: Bayes Theorem & Visualization
from PIL import Image, ImageDraw

def calculate_bayes(prior_D, sensitivity, fp_rate):
    p_pos = (sensitivity * prior_D) + (fp_rate * (1.0 - prior_D))
    post = (sensitivity * prior_D) / p_pos
    return post, p_pos

prior = 0.01
sens = 0.95
fp1 = 0.05
fp2 = 0.01

post1, _ = calculate_bayes(prior, sens, fp1)
post2, _ = calculate_bayes(prior, sens, fp2)

print("=" * 65)
print("EXERCISE 5 & 6: BAYES' THEOREM RESULTS")
print("=" * 65)
print(f"Scenario 1 (5% False Positives) : P(D | +) = {post1:.4f} ({post1*100:.2f}%)")
print(f"Scenario 2 (1% False Positives) : P(D | +) = {post2:.4f} ({post2*100:.2f}%)")

# Render Visualization Chart
width, height = 700, 400
img = Image.new('RGB', (width, height), color='#ffffff')
draw = ImageDraw.Draw(img)

margin_left, margin_bottom, plot_w, plot_h = 70, 60, 560, 280
for i in range(6):
    val = i / 5.0
    y_pixel = 50 + plot_h - int(val * plot_h)
    draw.line([(margin_left, y_pixel), (margin_left + plot_w, y_pixel)], fill='#e0e0e0')
    draw.text((margin_left - 40, y_pixel - 6), f"{val:.1f}", fill='#333333')

cats = ["Prior P(D)", "P(D|+) [FP=5%]", "P(D|+) [FP=1%]"]
vals = [prior, post1, post2]
cols = ['#1f77b4', '#ff7f0e', '#2ca02c']
col_w = plot_w // 3

for idx, (cat, val, col) in enumerate(zip(cats, vals, cols)):
    cx = margin_left + idx * col_w + col_w // 2
    bh = int(val * plot_h)
    draw.rectangle([(cx - 25, 50 + plot_h - bh), (cx + 25, 50 + plot_h)], fill=col)
    draw.text((cx - 15, 50 + plot_h - bh - 18), f"{val*100:.1f}%", fill='#111111')
    draw.text((cx - 45, 50 + plot_h + 10), cat, fill='#111111')

img.save("exercise6_bayes_chart.png")
print("Chart saved as exercise6_bayes_chart.png!")
```
