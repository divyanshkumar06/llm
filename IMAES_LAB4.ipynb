Observed facts

      ↓

Find applicable rules

      ↓

Fire rule

      ↓

Add new fact

      ↓

New rules may become applicable

      ↓

Repeat

      ↓

Identify animal

```python

# ============================================================
# EXPERIMENT 3
# FORWARD CHAINING INFERENCE ENGINE
# DATA-DRIVEN REASONING
# ============================================================


# ============================================================
# RULE BASE
# ============================================================

# Each rule contains:
# id          -> Rule identifier
# conditions  -> Facts required for the rule to fire
# conclusion  -> New fact derived after firing the rule


rules = [

    # --------------------------------------------------------
    # BASIC CLASSIFICATION RULES
    # --------------------------------------------------------

    {
        "id": "R1",
        "conditions": {"has hair"},
        "conclusion": "is mammal"
    },

    {
        "id": "R2",
        "conditions": {"gives milk"},
        "conclusion": "is mammal"
    },

    {
        "id": "R3",
        "conditions": {"has feathers"},
        "conclusion": "is bird"
    },

    {
        "id": "R4",
        "conditions": {"is mammal", "eats meat"},
        "conclusion": "is carnivore"
    },

    {
        "id": "R5",
        "conditions": {"has pointed teeth", "has claws"},
        "conclusion": "is carnivore"
    },


    # --------------------------------------------------------
    # ANIMAL IDENTIFICATION RULES
    # --------------------------------------------------------

    # CHEETAH
    {
        "id": "R6",
        "conditions": {
            "is mammal",
            "is carnivore",
            "has black spots"
        },
        "conclusion": "is cheetah"
    },


    # LEOPARD
    {
        "id": "R7",
        "conditions": {
            "is mammal",
            "is carnivore",
            "has tawny color",
            "has dark spots"
        },
        "conclusion": "is leopard"
    },


    # TIGER - EXERCISE 1
    {
        "id": "R8",
        "conditions": {
            "is mammal",
            "is carnivore",
            "has black stripes"
        },
        "conclusion": "is tiger"
    },


    # GIRAFFE - EXERCISE 1
    {
        "id": "R9",
        "conditions": {
            "is mammal",
            "has long neck",
            "has long legs"
        },
        "conclusion": "is giraffe"
    },


    # ZEBRA - EXERCISE 1
    {
        "id": "R10",
        "conditions": {
            "is mammal",
            "has black stripes",
            "not carnivore"
        },
        "conclusion": "is zebra"
    }

]


# ============================================================
# HELPER FUNCTION
# FIND ALL APPLICABLE RULES
# ============================================================

def find_applicable_rules(rules, facts):

    applicable = []

    for rule in rules:

        # Rule can fire when:
        #
        # 1. All conditions are present in facts
        #
        # 2. Conclusion is not already known

        if (
            rule["conditions"].issubset(facts)
            and rule["conclusion"] not in facts
        ):

            applicable.append(rule)

    return applicable


# ============================================================
# HELPER FUNCTION
# FIND MOST SPECIFIC CLASS
# ============================================================

def most_specific_class(facts):

    # Specificity order
    # More specific classes should be checked first

    specific_classes = [

        "is tiger",
        "is giraffe",
        "is zebra",
        "is cheetah",
        "is leopard",

        "is carnivore",
        "is bird",
        "is mammal"
    ]

    for animal_class in specific_classes:

        if animal_class in facts:

            return animal_class

    return "No classification established"


# ============================================================
# FORWARD CHAINING
# FIRE FIRST APPLICABLE RULE
# ============================================================

def forward_chaining_first_rule(
    rules,
    initial_facts,
    verbose=True
):

    # Working Memory
    facts = set(initial_facts)

    # Record fired rules
    firing_sequence = []

    cycle = 1

    if verbose:

        print("\n" + "=" * 60)
        print("FORWARD CHAINING - FIRST RULE STRATEGY")
        print("=" * 60)

        print("\nInitial Working Memory:")

        for fact in sorted(facts):
            print(" -", fact)


    # --------------------------------------------------------
    # FORWARD CHAINING LOOP
    # --------------------------------------------------------

    while True:

        # Step 1:
        # Find all applicable rules

        conflict_set = find_applicable_rules(
            rules,
            facts
        )


        if verbose:

            print("\n" + "-" * 60)

            print("Cycle:", cycle)

            print("Conflict Set:")

            if conflict_set:

                for rule in conflict_set:
                    print(
                        rule["id"],
                        "->",
                        rule["conclusion"]
                    )

            else:
                print("No applicable rules")


        # Step 2:
        # If no rule is applicable, stop

        if not conflict_set:
            break


        # Step 3:
        # Select first rule according
        # to the rule base order

        selected_rule = conflict_set[0]


        # Step 4:
        # Fire rule

        facts.add(
            selected_rule["conclusion"]
        )


        # Step 5:
        # Record rule

        firing_sequence.append(
            selected_rule["id"]
        )


        if verbose:

            print(
                "\nFiring Rule:",
                selected_rule["id"]
            )

            print(
                "New Fact:",
                selected_rule["conclusion"]
            )


        cycle += 1


    # --------------------------------------------------------
    # FINAL RESULT
    # --------------------------------------------------------

    identification = most_specific_class(
        facts
    )


    if verbose:

        print("\n" + "=" * 60)

        print("FINAL WORKING MEMORY")

        print("=" * 60)

        for fact in sorted(facts):
            print(" -", fact)


        print("\nFiring Sequence:")

        if firing_sequence:

            print(
                " -> ".join(firing_sequence)
            )

        else:

            print("No rules fired")


        print("\nMost Specific Classification:")

        print(identification)


    return {

        "facts": facts,

        "firing_sequence": firing_sequence,

        "identification": identification
    }


# ============================================================
# FORWARD CHAINING
# FIRE ALL APPLICABLE RULES PER CYCLE
# EXERCISE 3
# ============================================================

def forward_chaining_all_rules(
    rules,
    initial_facts,
    verbose=True
):

    # Working Memory

    facts = set(initial_facts)


    # Record fired rules

    firing_sequence = []


    cycle = 1


    if verbose:

        print("\n" + "=" * 60)

        print(
            "FORWARD CHAINING - "
            "ALL APPLICABLE RULES STRATEGY"
        )

        print("=" * 60)


        print("\nInitial Working Memory:")

        for fact in sorted(facts):

            print(" -", fact)


    # --------------------------------------------------------
    # FORWARD CHAINING LOOP
    # --------------------------------------------------------

    while True:


        # Find conflict set

        conflict_set = find_applicable_rules(
            rules,
            facts
        )


        if verbose:

            print("\n" + "-" * 60)

            print("Cycle:", cycle)

            print("Applicable Rules:")


            if conflict_set:

                for rule in conflict_set:

                    print(
                        rule["id"],
                        "->",
                        rule["conclusion"]
                    )

            else:

                print("No applicable rules")


        # Stop if no rule applies

        if not conflict_set:

            break


        # ----------------------------------------------------
        # FIRE ALL RULES
        # ----------------------------------------------------

        for rule in conflict_set:


            # Add conclusion

            facts.add(
                rule["conclusion"]
            )


            # Record rule

            firing_sequence.append(
                rule["id"]
            )


            if verbose:

                print(
                    "\nFiring:",
                    rule["id"]
                )

                print(
                    "Derived:",
                    rule["conclusion"]
                )


        cycle += 1


    # --------------------------------------------------------
    # FINAL RESULT
    # --------------------------------------------------------

    identification = most_specific_class(
        facts
    )


    if verbose:

        print("\n" + "=" * 60)

        print("FINAL WORKING MEMORY")

        print("=" * 60)


        for fact in sorted(facts):

            print(" -", fact)


        print("\nFiring Sequence:")

        if firing_sequence:

            print(
                " -> ".join(firing_sequence)
            )

        else:

            print("No rules fired")


        print(
            "\nMost Specific Classification:"
        )

        print(identification)


    return {

        "facts": facts,

        "firing_sequence":
            firing_sequence,

        "identification":
            identification
    }


# ============================================================
# EXERCISE 1
# IDENTIFY TIGER
# ============================================================

def test_tiger():

    print("\n\n")
    print("=" * 70)

    print("TEST 1: TIGER")

    print("=" * 70)


    observed_features = {

        "has hair",

        "eats meat",

        "has black stripes"
    }


    return forward_chaining_first_rule(

        rules,

        observed_features

    )


# ============================================================
# EXERCISE 1
# IDENTIFY GIRAFFE
# ============================================================

def test_giraffe():

    print("\n\n")
    print("=" * 70)

    print("TEST 2: GIRAFFE")

    print("=" * 70)


    observed_features = {

        "has hair",

        "has long neck",

        "has long legs"
    }


    return forward_chaining_first_rule(

        rules,

        observed_features

    )


# ============================================================
# EXERCISE 1
# IDENTIFY ZEBRA
# ============================================================

def test_zebra():

    print("\n\n")
    print("=" * 70)

    print("TEST 3: ZEBRA")

    print("=" * 70)


    observed_features = {

        "has hair",

        "has black stripes",

        "not carnivore"
    }


    return forward_chaining_first_rule(

        rules,

        observed_features

    )


# ============================================================
# EXERCISE 2
# NO ANIMAL IDENTIFICATION
# ============================================================

def test_partial_classification():

    print("\n\n")

    print("=" * 70)

    print(
        "TEST 4: PARTIAL CLASSIFICATION"
    )

    print("=" * 70)


    # These facts can establish:
    #
    # has hair
    #      ↓
    # is mammal
    #
    # But cannot identify a specific animal.


    observed_features = {

        "has hair"

    }


    return forward_chaining_first_rule(

        rules,

        observed_features

    )


# ============================================================
# MAIN
# ============================================================

if __name__ == "__main__":


    # --------------------------------------------------------
    # MAIN EXPERIMENT
    # --------------------------------------------------------

    print("\n\n")

    print("=" * 70)

    print(
        "MAIN EXPERIMENT: "
        "CHEETAH IDENTIFICATION"
    )

    print("=" * 70)


    # Initial observations

    observed_features = {

        "has hair",

        "eats meat",

        "has black spots"
    }


    # Fire FIRST applicable rule
    # per cycle

    main_result = (
        forward_chaining_first_rule(

            rules,

            observed_features

        )
    )


    # --------------------------------------------------------
    # EXERCISE 1
    # --------------------------------------------------------

    test_tiger()

    test_giraffe()

    test_zebra()


    # --------------------------------------------------------
    # EXERCISE 2
    # --------------------------------------------------------

    test_partial_classification()


    # --------------------------------------------------------
    # EXERCISE 3
    # COMPARE FIRST RULE VS ALL RULES
    # --------------------------------------------------------

    print("\n\n")

    print("=" * 70)

    print(
        "EXERCISE 3: "
        "FIRST RULE VS ALL APPLICABLE RULES"
    )

    print("=" * 70)


    comparison_facts = {

        "has hair",

        "eats meat",

        "has pointed teeth",

        "has claws",

        "has black spots"
    }


    print("\n\n")

    print(
        "STRATEGY 1: "
        "FIRE FIRST APPLICABLE RULE"
    )


    first_result = (
        forward_chaining_first_rule(

            rules,

            comparison_facts

        )
    )


    print("\n\n")

    print(
        "STRATEGY 2: "
        "FIRE ALL APPLICABLE RULES"
    )


    all_result = (
        forward_chaining_all_rules(

            rules,

            comparison_facts

        )
    )


    # --------------------------------------------------------
    # COMPARISON
    # --------------------------------------------------------

    print("\n\n")

    print("=" * 70)

    print("FINAL COMPARISON")

    print("=" * 70)


    print(
        "\nFirst Rule Strategy:"
    )

    print(
        " -> ".join(
            first_result[
                "firing_sequence"
            ]
        )
    )


    print(
        "\nAll Applicable Rules Strategy:"
    )

    print(
        " -> ".join(
            all_result[
                "firing_sequence"
            ]
        )
    )


    print(
        "\nFirst Strategy Identification:"
    )

    print(
        first_result[
            "identification"
        ]
    )


    print(
        "\nAll Rules Strategy Identification:"
    )

    print(
        all_result[
            "identification"
        ]
    )

```
